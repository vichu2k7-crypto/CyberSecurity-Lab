```python
# Create a sample login log file
with open("login_log.txt", "w") as file:
    file.write("192.168.1.10 FAILED\n")
    file.write("192.168.1.20 SUCCESS\n")
    file.write("192.168.1.10 FAILED\n")
    file.write("10.0.0.5 FAILED\n")
    file.write("192.168.1.10 FAILED\n")
    file.write("10.0.0.5 SUCCESS\n")

# Dictionary to store failed login counts
failed_attempts = {}

# Read the log file
with open("login_log.txt", "r") as file:
    for line in file:
        ip, status = line.strip().split()

        if status == "FAILED":
            if ip in failed_attempts:
                failed_attempts[ip] += 1
            else:
                failed_attempts[ip] = 1

# Display the results
print("Failed Login Attempts Per IP:\n")

for ip, count in failed_attempts.items():
    print(f"{ip} : {count} failed attempt(s)")
```

Title:
Write a Python program to parse a login log file and count failed login attempts per IP address.

Aim:
To parse a login log file and count the number of failed login attempts from each IP address.

Theory:
Log analysis is an important part of network security. Login logs contain records of successful and failed login attempts. By analyzing these logs, security administrators can identify suspicious activities such as repeated failed logins from the same IP address, which may indicate a brute-force attack.

Algorithm:
1. Create a sample login log file.
2. Read the log file line by line.
3. Check for failed login entries.
4. Extract the IP address from each failed login.
5. Count the number of failed login attempts for each IP address.
6. Display the results.

Result:
The program successfully parsed the login log file and counted the failed login attempts for each IP address. This demonstrates basic log analysis used in network security.
