```python
# List of blocked IP addresses
blocked_ips = [
    "192.168.1.100",
    "10.0.0.5",
    "172.16.0.20"
]

# Get packet details
source_ip = input("Enter Source IP Address: ")
destination_port = int(input("Enter Destination Port: "))

# Firewall rules
if source_ip in blocked_ips:
    print("\nFirewall Decision: BLOCKED")
    print("Reason: Source IP is in the blocked list.")
else:
    print("\nFirewall Decision: ALLOWED")
    print(f"Packet allowed to Destination Port {destination_port}.")
```

Result:
The program successfully simulated a firewall by checking the source IP address against predefined firewall rules and deciding whether to allow or block the network packet.

Title:
Write a Python program to simulate a firewall's decision to allow or block a network packet based on a set of rules.

Aim:
To simulate the working of a firewall by allowing or blocking network packets based on predefined rules.

Theory:
A firewall is a network security system that monitors and controls incoming and outgoing network traffic based on predefined security rules. It decides whether to allow or block a packet depending on parameters such as source IP address, destination port, or protocol.

Algorithm:
1. Create a list of blocked IP addresses.
2. Read the source IP address and destination port from the user.
3. Check whether the source IP is in the blocked list.
4. If the IP is blocked, deny access.
5. Otherwise, allow the packet.
6. Display the firewall decision.
