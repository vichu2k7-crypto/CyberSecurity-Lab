```python
# List of known malicious IP addresses (IoCs)
malicious_ips = [
    "192.168.1.10",
    "10.0.0.5",
    "172.16.0.20"
]

# Sample traffic log
traffic_log = [
    "192.168.1.2",
    "192.168.1.10",
    "8.8.8.8",
    "10.0.0.5",
    "172.16.0.25",
    "1.1.1.1"
]

print("Scanning Traffic Log...\n")

for ip in traffic_log:
    if ip in malicious_ips:
        print(f"ALERT: {ip} is a Malicious IP Address!")
    else:
        print(f"{ip} is Safe.")
```

Result:
The program successfully scanned the sample traffic log and identified IP addresses that matched the list of known malicious Indicators of Compromise (IoCs).

Title:
Write a Python program to scan a sample traffic log against a list of known malicious IP addresses (IoCs) and flag matches.

Aim:
To identify malicious IP addresses in a network traffic log using Indicators of Compromise (IoCs).

Theory:
Indicators of Compromise (IoCs) are known malicious IP addresses, domains, or file hashes that indicate suspicious activity. By comparing network traffic logs against a list of IoCs, security analysts can quickly detect potential threats.

Algorithm:
1. Create a list of known malicious IP addresses.
2. Create a sample traffic log containing IP addresses.
3. Compare each IP address in the traffic log with the IoC list.
4. If a match is found, display an alert.
5. Otherwise, mark the IP address as safe.
