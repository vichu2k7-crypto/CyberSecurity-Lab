```python
import hashlib

# Function to calculate SHA-256 hash
def calculate_hash(filename):
    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()

# Create a sample file
with open("sample.txt", "w") as file:
    file.write("Threat Intelligence Lab")

# Calculate original hash
original_hash = calculate_hash("sample.txt")
print("Original Hash:")
print(original_hash)

# Modify the file (simulate tampering)
with open("sample.txt", "a") as file:
    file.write("\nFile Modified")

# Calculate new hash
new_hash = calculate_hash("sample.txt")
print("\nNew Hash:")
print(new_hash)

# Compare hashes
if original_hash == new_hash:
    print("\nFile Integrity Verified. No Tampering Detected.")
else:
    print("\nWarning! File Tampering Detected.")
```

Result:
The program successfully calculated the SHA-256 hash of a file before and after modification. Since the hash values were different, the program correctly detected file tampering, demonstrating the concept of data integrity.
