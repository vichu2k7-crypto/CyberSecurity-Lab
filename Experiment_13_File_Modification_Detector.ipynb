```python
import hashlib

# Function to calculate SHA-256 hash
def calculate_hash(filename):
    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()

# Create a sample file
filename = "sample_file.txt"

with open(filename, "w") as file:
    file.write("This is the original file content.")

# Calculate baseline hash
baseline_hash = calculate_hash(filename)
print("Baseline Hash:")
print(baseline_hash)

# Modify the file
with open(filename, "a") as file:
    file.write("\nThis line was added later.")

# Calculate current hash
current_hash = calculate_hash(filename)
print("\nCurrent Hash:")
print(current_hash)

# Compare hashes
print("\nVerification Result:")
if baseline_hash == current_hash:
    print("File is NOT modified.")
else:
    print("File has been MODIFIED.")
```

Result:
The program successfully compared the current SHA-256 hash of the file with the known-good baseline hash and correctly detected that the file had been modified.

Title:
Write a Python program to detect whether a file has been modified by comparing its hash against a known-good baseline.

Aim:
To detect whether a file has been modified by comparing its current SHA-256 hash with a known-good baseline hash.

Theory:
A cryptographic hash function generates a unique hash value for a file. A known-good baseline hash is the original hash of the file. If the file is modified, its hash changes. Comparing the current hash with the baseline helps detect unauthorized modifications and ensures file integrity.

Algorithm:
1. Import the hashlib module.
2. Create a sample file.
3. Calculate and store its baseline SHA-256 hash.
4. Modify the file.
5. Calculate the current SHA-256 hash.
6. Compare the current hash with the baseline hash.
7. Display whether the file has been modified.
