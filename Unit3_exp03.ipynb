```python
# Experiment 3 Setup
import statistics

print("Setup completed successfully.")
```

```python
# Experiment 3: Hypothesis-, Intelligence-, and Analytics-Based Threat Hunting

# --------------------------------------------------
# 1. Hypothesis-Based Threat Hunting
# --------------------------------------------------

# Hypothesis:
# Attackers may use encoded PowerShell commands

execution_logs = [
    {
        "process": "powershell.exe",
        "args": "-enc SGVsbG8=",
        "user": "svc_admin"
    },
    {
        "process": "notepad.exe",
        "args": "",
        "user": "jdoe"
    },
    {
        "process": "powershell.exe",
        "args": "-nop -w hidden -enc ZXZpbA==",
        "user": "jdoe"
    }
]


def hypothesis_hunt_powershell(logs):
    return [
        l for l in logs
        if l["process"] == "powershell.exe"
        and "-enc" in l["args"]
    ]


# --------------------------------------------------
# 2. Intelligence-Based Threat Hunting
# --------------------------------------------------

# Match connections against a known-malicious IP feed

connection_logs = [
    {"ip": "45.33.32.156"},
    {"ip": "8.8.8.8"},
    {"ip": "185.220.101.1"}
]

known_malicious_ips = {
    "45.33.32.156",
    "185.220.101.1"
}


def intelligence_hunt(logs, feed):
    return [
        l for l in logs
        if l["ip"] in feed
    ]


# --------------------------------------------------
# 3. Analytics-Based Threat Hunting
# --------------------------------------------------

# Normal login hours are approximately 8 AM - 10 AM

login_hours_history = [9, 9, 10, 9, 8, 9, 10]
new_login_hour = 3


def analytics_hunt(history, new_value, threshold=2.0):
    mean = statistics.mean(history)
    stdev = statistics.pstdev(history) or 1

    z = abs(new_value - mean) / stdev

    return {
        "z_score": round(z, 2),
        "anomalous": z > threshold
    }


# --------------------------------------------------
# Execute all three hunting techniques
# --------------------------------------------------

hyp_result = hypothesis_hunt_powershell(execution_logs)

intel_result = intelligence_hunt(
    connection_logs,
    known_malicious_ips
)

analytics_result = analytics_hunt(
    login_hours_history,
    new_login_hour
)


# Display results

print("Hypothesis-based hits:", hyp_result)
print("Intelligence-based hits:", intel_result)
print("Analytics-based result:", analytics_result)
```

```python
# Test Cases for Experiment 3

def test_experiment3():

    # Hypothesis-based test
    assert len(hyp_result) == 2, \
        "Both encoded PowerShell executions should be flagged"

    assert all(
        r["process"] == "powershell.exe"
        for r in hyp_result
    )


    # Intelligence-based test
    assert len(intel_result) == 2, \
        "Both known-malicious IPs should be matched"


    # Analytics-based test
    assert analytics_result["anomalous"] is True, \
        "A 3 AM login should be anomalous against an 8-10 AM history"


    print("Experiment 3: All test cases passed.")


test_experiment3()
```
