```python
import re

# Get URL from the user
url = input("Enter a URL: ")

score = 0

# Check for IP address in URL
if re.search(r'https?://\d+\.\d+\.\d+\.\d+', url):
    print("Warning: URL contains an IP address.")
    score += 1

# Check for '@' symbol
if '@' in url:
    print("Warning: URL contains '@' symbol.")
    score += 1

# Check for long URL
if len(url) > 75:
    print("Warning: URL is unusually long.")
    score += 1

# Check for multiple hyphens
if url.count('-') >= 2:
    print("Warning: URL contains multiple hyphens.")
    score += 1

# Final result
if score > 0:
    print("\nResult: Suspicious URL (Possible Phishing)")
else:
    print("\nResult: URL appears safe.")
```

Result:
The program successfully detected common phishing indicators in a URL and classified it as either suspicious or likely safe.

Title:
Write a Python program to flag a suspicious URL as a likely phishing link based on common phishing indicators.

Aim:
To identify suspicious URLs based on common phishing characteristics.

Theory:
Phishing websites often use deceptive URLs to trick users into revealing sensitive information. Common phishing indicators include the use of IP addresses instead of domain names, the '@' symbol, excessive hyphens, and unusually long URLs. This program checks these indicators and warns the user if the URL appears suspicious.

Algorithm:
1. Read the URL from the user.
2. Check for common phishing indicators.
3. Count the number of suspicious indicators.
4. If one or more indicators are found, mark the URL as suspicious.
5. Otherwise, mark the URL as likely safe.
