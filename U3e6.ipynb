```python
# Experiment 6: Continuous Security Monitoring with Anomaly Flagging

import random

# Fix random seed so results are reproducible
random.seed(42)


# Generate simulated event stream
def generate_event_stream(n=20):
    stream = []

    for _ in range(n):
        stream.append({
            "hour": random.choice([9, 10, 11, 14, 15, 16]),
            "failed_logins": random.choice([0, 0, 0, 1])
        })

    # Inject two known anomalies
    stream.append({
        "hour": 3,
        "failed_logins": 0
    })

    stream.append({
        "hour": 10,
        "failed_logins": 9
    })

    return stream


# Monitor events
def monitor_stream(stream):
    flags = []

    for event in stream:
        reasons = []

        # Detect off-hours activity
        if event["hour"] < 6 or event["hour"] > 22:
            reasons.append("off-hours activity")

        # Detect excessive failed logins
        if event["failed_logins"] >= 5:
            reasons.append("excessive failed logins")

        if reasons:
            flags.append({
                "event": event,
                "reasons": reasons
            })

    return flags


# Generate and monitor stream
event_stream = generate_event_stream()

flags = monitor_stream(event_stream)


# Display results
print(
    f"Stream length: {len(event_stream)} | "
    f"Flags raised: {len(flags)}"
)

for f in flags:
    print(f)
```

```python
# Test Cases for Experiment 6

def test_experiment6():

    assert len(flags) == 2, \
        "Exactly the 2 injected anomalies should be flagged"

    assert any(
        "off-hours activity" in f["reasons"]
        for f in flags
    )

    assert any(
        "excessive failed logins" in f["reasons"]
        for f in flags
    )

    print("Experiment 6: All test cases passed.")


test_experiment6()
```
