```python
import hashlib

# Original password
password = "security123"

# Generate SHA-256 hash
target_hash = hashlib.sha256(password.encode()).hexdigest()

print("Target Hash:")
print(target_hash)

# Sample wordlist
wordlist = [
    "admin",
    "password",
    "welcome",
    "security123",
    "123456",
    "qwerty"
]

print("\nStarting Dictionary Attack...\n")

found = False

for word in wordlist:
    word_hash = hashlib.sha256(word.encode()).hexdigest()

    if word_hash == target_hash:
        print("Password Found!")
        print("Password:", word)
        found = True
        break

if not found:
    print("Password Not Found in Wordlist.")
```

Result:
The program successfully demonstrated a dictionary attack by comparing SHA-256 hashes of passwords in a wordlist with the target hash. When a matching hash was found, the original password was identified.

Title:
Write a Python program to crack a hashed password using a wordlist, demonstrating a dictionary attack.

Aim:
To demonstrate a dictionary attack by comparing password hashes with a wordlist.

Theory:
A dictionary attack is a password cracking technique where a list of common passwords (wordlist) is tested against a target password hash. Each word is hashed using the same algorithm, and the hashes are compared. If a match is found, the password is identified.

Algorithm:
1. Import the hashlib module.
2. Create a sample password.
3. Generate its SHA-256 hash.
4. Create a wordlist containing possible passwords.
5. Hash each word in the wordlist.
6. Compare each hash with the target hash.
7. Display the password if a match is found.
