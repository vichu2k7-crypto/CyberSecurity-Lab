```python
# XOR Encryption Function
def xor_encrypt_decrypt(text, key):
    result = ""

    for char in text:
        result += chr(ord(char) ^ key)

    return result

# Secret key
key = 15

# User input
message = input("Enter the Message: ")

# Encryption
encrypted = xor_encrypt_decrypt(message, key)
print("\nEncrypted Message:")
print(encrypted)

# Decryption
decrypted = xor_encrypt_decrypt(encrypted, key)
print("\nDecrypted Message:")
print(decrypted)
```

Result:
The program successfully simulated secure VPN-like communication using symmetric key encryption. The original message was encrypted and then correctly decrypted using the same secret key.

Title:
Write a Python program to simulate secure, VPN-like communication between two parties using symmetric key encryption.

Aim:
To simulate secure communication between two parties using symmetric key encryption.

Theory:
A Virtual Private Network (VPN) provides secure communication by encrypting data before transmission and decrypting it at the receiving end. Symmetric key encryption uses the same secret key for both encryption and decryption. This experiment demonstrates the basic concept of secure communication using a simple XOR-based symmetric encryption technique.

Algorithm:
1. Read the message from the user.
2. Define a secret key.
3. Encrypt the message using XOR encryption.
4. Display the encrypted message.
5. Decrypt the encrypted message using the same key.
6. Display the original message.
