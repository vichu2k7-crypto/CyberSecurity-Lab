```python
import socket

# Get target host
host = input("Enter Host (Example: scanme.nmap.org or localhost): ")

# Get port range
start_port = int(input("Enter Starting Port: "))
end_port = int(input("Enter Ending Port: "))

print("\nScanning", host)
print("-" * 30)

for port in range(start_port, end_port + 1):
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(1)

    result = s.connect_ex((host, port))

    if result == 0:
        print(f"Port {port} is OPEN")

    s.close()

print("\nScanning Completed.")
```

Result:
The Python program successfully scanned the specified TCP port range and displayed the open ports on the target host.

Title:
Write a Python program to scan a range of TCP ports on a target host and report which ports are open.

Aim:
To identify open TCP ports on a target host using Python.

Theory:
Port scanning is a technique used to identify open ports on a computer or server. Open ports indicate that a service is running and listening for network connections. This helps administrators monitor network security and identify available services.

Algorithm:
1. Import the socket module.
2. Read the target host name or IP address.
3. Read the starting and ending port numbers.
4. Connect to each port using TCP.
5. If the connection is successful, report the port as open.
6. Display all open ports found.
