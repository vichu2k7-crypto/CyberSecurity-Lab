```python
import socket

# List of domain names
domains = [
    "google.com",
    "github.com",
    "openai.com",
    "example.com"
]

print("Domain Name\t\tIP Address")
print("-" * 45)

for domain in domains:
    try:
        ip = socket.gethostbyname(domain)
        print(f"{domain}\t\t{ip}")
    except socket.gaierror:
        print(f"{domain}\t\tUnable to resolve")
```

Result:
The program successfully resolved the given domain names to their corresponding IP addresses using Python's socket module, demonstrating a basic OSINT technique.
