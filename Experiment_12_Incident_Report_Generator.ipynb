```python
# List of detected security events
security_events = [
    "Multiple failed login attempts from 192.168.1.10",
    "Malicious IP detected: 10.0.0.5",
    "Suspicious URL identified: http://fake-bank-login.com",
    "Port scan detected from 172.16.0.20"
]

print("=" * 50)
print("        SECURITY INCIDENT REPORT")
print("=" * 50)

for i, event in enumerate(security_events, start=1):
    print(f"{i}. {event}")

print("\nTotal Security Events:", len(security_events))
print("Report Status: Generated Successfully")
print("=" * 50)
```

Result:
The program successfully generated a structured security incident report from the detected security events and displayed the total number of incidents.

Title:
Write a Python program to automatically generate a structured incident report from a list of detected security events.

Aim:
To generate a structured incident report from a list of detected security events using Python.

Theory:
An incident report is a document that summarizes security-related events detected in a system or network. It helps security analysts understand the type of incident, the affected system, and the recommended actions. Automating incident report generation improves efficiency and ensures consistent reporting.

Algorithm:
1. Create a list of detected security events.
2. Display the incident report header.
3. Print each security event with a serial number.
4. Display the total number of detected events.
5. Display the report generation status.
