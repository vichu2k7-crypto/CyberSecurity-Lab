```python
# Sample login log
login_log = [
    ("192.168.1.10", "FAILED"),
    ("192.168.1.20", "SUCCESS"),
    ("192.168.1.10", "FAILED"),
    ("10.0.0.5", "FAILED"),
    ("192.168.1.10", "FAILED"),
    ("10.0.0.5", "FAILED"),
    ("10.0.0.5", "FAILED")
]

# Threshold for alert
threshold = 3

failed_attempts = {}

# Count failed logins
for ip, status in login_log:
    if status == "FAILED":
        failed_attempts[ip] = failed_attempts.get(ip, 0) + 1

print("Intrusion Detection Report")
print("-" * 30)

# Check for suspicious activity
for ip, count in failed_attempts.items():
    print(f"{ip} : {count} failed login(s)")
    if count >= threshold:
        print(f" ALERT! Possible Brute-Force Attack Detected from {ip}\n")
```

Result:
The program successfully implemented a basic signature-based Intrusion Detection System (IDS). It detected repeated failed login attempts and generated alerts when the number of failed logins reached the specified threshold.

Title:
Write a Python program implementing a basic signature-based Intrusion Detection System that raises an alert on repeated failed logins.

Aim:
To implement a basic signature-based Intrusion Detection System (IDS) that detects repeated failed login attempts.

Theory:
A Signature-Based Intrusion Detection System (IDS) identifies malicious activities by comparing events with predefined attack signatures. In this experiment, repeated failed login attempts from the same IP address are treated as a known attack pattern. If the number of failed attempts exceeds a threshold, an alert is generated.

Algorithm:
1. Create a sample login log.
2. Count the failed login attempts for each IP address.
3. Set a threshold value (e.g., 3 failed attempts).
4. Compare the failed login count with the threshold.
5. Generate an alert if the threshold is reached.
6. Display the results.
